Start by importing the pipeline file, here lies the function that automates the process to generate the different diagrams. 

After we need to define the parameters used:
- **theory**: indicate the theory used in the calculations (e.g. "phi4", "qcd", "qcd_gluons")
- **in_particles**: indicate the number of particles of a certain type that goes in to the diagram. (e.g. [1, 2] indicate 1 particles of type a certain type and 2 particles of the other type). In the case of theories with one type of particles an integer can be used.
- **out_particles**: similarly indicate the number of particles of a certain type coming out of the diagram, following the same logic as **in_particles**.
- **order**: indicate the order in the perturbative expansion to calculate.

After indicating the parameters the program offers 2 option:

- **complete**: the program return all diagrams at the specified order that contribute to the process considered.
- **framework**: indicate the combination of initial canonical diagrams that contribute to the process at said order. With later the option of generating the final diagrams that comes from said canonical diagrams.

```python
from rgpep_dgt import calculate_diagrams, represent_order_diagram

theory = "phi4" # The theory to calculate diagrams for, e.g., "phi4", "qcd", "qcd_gluons"
in_particles = 1 # Number of incoming particles, for each type of particles
out_particles = 1 # Number of outgoing particles, for each type of particles
order = 7 # The order of the perturbation theory to calculate diagrams

mode = "framework" # The mode to calculate diagrams, e.g., "complete" or "framework"
```

```python
points, connections, count, in_out = calculate_diagrams(theory, in_particles, out_particles, order, mode)
```

```python
represent_order_diagram(points, connections, count, in_out, mode, order, process=[in_particles, out_particles])
```
